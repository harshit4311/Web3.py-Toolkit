```python
from web3 import Web3
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
infura_pid = os.getenv("INFURA_PROJECT_ID")
```

```python
infura_url = f'https://mainnet.infura.io/v3/{infura_pid}'
web3 = Web3(Web3.HTTPProvider(infura_url))
print("Connected:", web3.is_connected())
```

#### significance of --> "Connected: True"

What This Really Means:
1. Your code is talking to a live Ethereum node.

2. You can read data from the blockchain (balances, blocks, contract state, etc).

3. You can send transactions, including:

- ETH transfers

- Interacting with smart contracts (e.g., NFTs, DeFi protocols)

- Deploying new contracts

```python
address = '0x742d35Cc6634C0532925a3b844Bc454e4438f44e' # some random wallet address i came across on Etherscan
balance = web3.eth.get_balance(address)
eth_balance = web3.from_wei(balance, 'ether')
print(f"Balance: {eth_balance} ETH")
```

```python
latest_block = web3.eth.get_block('latest')
print(latest_block)
```

```python
"in better readable format"

import json

latest_block = web3.eth.get_block('latest')

def clean_dict(data):
    if isinstance(data, bytes):
        return data.hex()
    elif isinstance(data, list):
        return [clean_dict(item) for item in data]
    elif isinstance(data, dict):
        return {k: clean_dict(v) for k, v in data.items()}
    else:
        return data

cleaned_block = clean_dict(dict(latest_block))

print(json.dumps(cleaned_block, indent=2))
```
